```python
from transformers import pipeline
```

```python
# Load the classification pipeline with the specified model
model = pipeline("text-classification", model="tabularisai/multilingual-sentiment-analysis")
```

```python
# Classify a new sentence
review = """
I have joined Full stack Data Science program under Siva Rama Krishna sir. 
some concepts explained very clearly and some of them were not clear.
The entire course content is very clear and very helpful for me to gain knowledge.
Srk sir is also explained all the concepts very clearly. It's beneficial for my career.
but if he add more projects, that would be better.
"""


# Print the result
result = model(review)
print(result)
```

```python
from transformers import pipeline
classifier = pipeline('zero-shot-classification', model='roberta-large-mnli')
```

```python
review = """
I have joined Full stack Data Science program under Siva Rama Krishna sir. 
some concepts explained very clearly and some of them were not clear.
The entire course content is very clear and very helpful for me to gain knowledge.
Srk sir is also explained all the concepts very clearly. It's beneficial for my career.
but if he add more projects, that would be better.
"""

candidate_labels = ['travel', 'posive', 'very positive', 'negative']

classifier(review, candidate_labels)['labels'][0]
```
