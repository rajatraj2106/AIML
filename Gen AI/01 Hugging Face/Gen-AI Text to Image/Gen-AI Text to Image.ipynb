```python
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt
import torch
```

```python
# Load model
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to("cuda")
```

```python
# Prompt
prompt = (
    "A beautiful scenic location with a happy 12-year-old boy named Reyansh Raj, "
    "standing near a crystal-clear lake surrounded by green mountains, colorful flowers, "
    "bright blue sky, cinematic lighting, ultra realistic, highly detailed, 8k"
)

# Generate image
result = pipe(prompt)

# Extract PIL image
image = result.images[0]

# Display
plt.figure(figsize=(8, 8))
plt.imshow(image)
plt.axis("off")
plt.show()
```
