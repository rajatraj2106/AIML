```python
ARTICLE = """ I had an excellent experience! 
              The quality exceeded my expectations, and the attention to detail was impressive. 
              Everything was handled professionally, and the customer service was friendly, responsive, and truly went above and beyond. 
              I’m extremely satisfied and would definitely choose this again. Highly recommended!
"""
```

```python
from transformers import pipeline
```

```python
# Load the summarization pipeline with the specified model
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")
```

```python
print(summarizer(ARTICLE, max_length=100, min_length=70, do_sample=False))
```
